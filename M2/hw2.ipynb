## Section 4 — Type correction

```python
# This cell creates a version of the dataset where avg_temp_c was
# accidentally saved as text — a common real-world loading error.
df_broken = df.copy()
df_broken["avg_temp_c"] = df_broken["avg_temp_c"].astype(str)
```

```python
# 4a. Confirm that avg_temp_c is now stored as object (text).
#     Print just the dtype of that one column.
print("Broken dtype:", df_broken["avg_temp_c"].dtypes)

# 4b. Fix it: convert avg_temp_c back to float.
df_broken["avg_temp_c"] = df_broken["avg_temp_c"].astype(float)

# 4c. Confirm the fix. It should show float64.
print("Fixed dtype:", df_broken["avg_temp_c"].dtypes)
```
