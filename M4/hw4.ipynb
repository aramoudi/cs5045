**Dataset:** `data.csv`: 15 books, columns: `title`, `author`, `genre`, `year`, `pages`, `available`

## Setup

Run this cell first. It loads the dataset. All later sections depend on `books`.

```python
import csv

books = []
with open('data.csv') as f:
    reader = csv.DictReader(f)
    for row in reader:
        row['year']      = int(row['year'])
        row['pages']     = int(row['pages'])
        row['available'] = (row['available'] == 'True')
        books.append(row)

print(f'Loaded {len(books)} books.')
print('Columns:', list(books[0].keys()))
```

## Section 1: Load and Navigate

Confirm the dataset loaded correctly and practice index-based access.

```python
# 1a. Print the total number of books loaded.
#     Expected: 15
print('Total books:', len(books))

# 1b. Print the title and year of the FIRST book.
print('First book:', books[0]['title'], books[0]['year'])

# 1c. Print the title and year of the LAST book.
#     Hint: use a negative index.
print('Last book:', books[-1]['title'], books[-1]['year'])
```

## Section 2: List Indexing and Slicing

Practice building new lists from the data using comprehensions and slices.

```python
# 2a. Build a list called all_titles containing ONLY the title of every book,
#     in the same order as books. Use a list comprehension.
all_titles = [book['title'] for book in books]
print('All titles:', all_titles)

# 2b. Build a list called available_titles containing the titles of books
#     where available is True. Use a list comprehension with an if condition.
available_titles = [book['title'] for book in books if book['available']]
print('Available titles:', available_titles)

# 2c. Print the titles of the FIRST THREE books using a slice.
print('First three titles:', [b['title'] for b in books[:3]])
```

## Section 3: Dictionary Lookup

Build a lookup index so you can find any book by title without looping.

```python
# 3a. Build a dictionary called catalog that maps each title (str)
#     to the book's full dictionary. Use a dictionary comprehension.
catalog = {book['title']: book  for book in books}

# 3b. Use the catalog to look up "Factfulness" and print the author's name.
factfulness = catalog['Factfulness']
print('Factfulness author:', factfulness['author'])

# 3c. Use the catalog to look up "Bad Science" and print its page count.
print('Bad Science pages:', catalog['Bad Science']['pages'])
```

## Section 4: Transform, Adding Computed Columns

Add new fields to every book dictionary without changing the existing ones.

```python
# 4a. Add a 'decade' key to every book.
#     decade = (year // 10) * 10
#     Example: 2012 -> 2010,  2008 -> 2000
for book in books:
    book['decade'] = (book['year'] // 10) * 10

print('Decade check (The Signal and the Noise):', books[0]['decade'])
# Expected: 2010

# 4b. Add a 'long' key to every book.
#     long = True when pages > 500, False otherwise.
for book in books:
    book['long'] = book['pages'] > 500

print('Long-book check (The Signal and the Noise):', books[0]['long'])
# Expected: True  (534 pages)
print('Long-book check (Thinking Fast and Slow):', books[1]['long'])
# Expected: False (499 pages)

# 4c. Build a list called long_titles containing the titles of all long books.
#     Use a list comprehension.
long_titles = [book['title'] for book in books if book['long']]
print('Long books (>500 pages):', long_titles)
```

## Section 5: Grouping and Tuples

Group books by genre and practice creating tuple records.

```python
# 5a. Build a dictionary called by_genre that maps each genre (str)
#     to a list of titles in that genre.
#     Structure: { "Statistics": ["The Signal...", "The Art of..."], ... }
by_genre = {}
for book in books:
    g = book['genre']
    if g not in by_genre:
        by_genre[g] = []
    by_genre[g].append(book['title'])   # append the book's title

print('Distinct genres:', len(by_genre))
print('Statistics books:', by_genre['Statistics'])

# 5b. Create a tuple storing (title, author, year) of the LAST book.
#     Assign it to last_record.
last_record = (books[-1]['title'], books[-1]['author'], books[-1]['year'])
print('Last record tuple:', last_record)
print('Type of last_record:', type(last_record))
```
