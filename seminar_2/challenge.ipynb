## Data Loading (pre-written; run this cell first)

This cell loads `challenge_data.csv` into a variable called `rows`.  
Each element of `rows` is a **dictionary**, one per US state.  
Keys are column names; values are that state's data.

After running this cell you will see: `Dataset loaded: 51 rows`

```python
import csv
import os

DATA_PATH = 'challenge_data.csv'

rows = []
with open(DATA_PATH, newline='') as f:
    reader = csv.DictReader(f)
    for row in reader:
        row['obesity_pct']     = float(row['obesity_pct'])
        row['smoking_pct']     = float(row['smoking_pct'])
        row['exercise_pct']    = float(row['exercise_pct'])
        row['life_expectancy'] = float(row['life_expectancy'])
        row['median_income']   = int(row['median_income'])
        rows.append(row)

print(f'Dataset loaded: {len(rows)} rows')
print('Columns:', list(rows[0].keys()))
```

List all unique values in the `region` column. Use a loop; if you have not seen this region yet, add it to your list.

### Task 4: Filter

Keep only the states where `obesity_pct` is greater than 33.  
Store the matching rows in a list. Print how many states remain and their names.

```python
# Task 4
states = []
for row in rows:
    if row['obesity_pct'] > 33:
        if row['state'] not in states:
            print(f"{row['state']} ({row['region']}): {row['obesity_pct']}%")
            states.append(row)

print(f"\nNumber of states with high obesity: {len(states)} states")
```

## Part B: Transform and Aggregate

Tasks 5–8. You are computing new values, collecting subsets, and summarizing by group.

### Task 5: Transform

For every row in `rows`, add a new key called `smoker_level`.  
If `smoking_pct` is greater than 20, set it to `'high_smoker'`. Otherwise, `'low_smoker'`.

After the loop, print the first 5 rows showing `state` and `smoker_level` to verify.

**Expected:** 14 states classified as `high_smoker`.

```python
# Task 5
new_ds = []
count_of_high_smoker = 0
for row in rows:
    if row['smoking_pct'] > 20:
        row['smoker_level'] = 'high_smoker'
        count_of_high_smoker += 1
    else:
        row['smoker_level'] = 'low_smoker'
    new_ds.append(row)

for i in range(5):
    print(f"Row #{i+1}")
    print('\t', 'state: \t', rows[i]['state'])
    print('\t', 'smoker_level: \t', rows[i]['smoker_level'])
print("... printed only 5 rows\n")

print(f"{count_of_high_smoker} states classified as `smoker_level`")
```

### Task 6: Collect: Filter

Collect the names of all states where `life_expectancy` is greater than 79.  
Store the state names (not the full row) in a list. Print the list and the count.

**Expected:** 16 states.

```python
# Task 6
states = []
for row in rows:
    if row['life_expectancy'] > 79:
        states.append(row['state'])

print(f"List of States: {', '.join(states)}")
print(f"Total of {len(states)} states")
```

### Task 7: Aggregate by Region

Compute the mean `obesity_pct` for each census region.

**Strategy:** This takes two passes.
1. First, build a dictionary where each key is a region name and each value is a list of `obesity_pct` values for states in that region.
2. Then loop over that dictionary and compute `sum(values) / len(values)` for each region.

**Expected (approximately):** South ~33.6%, Midwest ~32.6%, West ~27.8%, Northeast ~27.7%.

```python
# Task 7

def compute_mean(ds, key):
    return round(sum(ds[key])/len(ds[key]), 1)

new_ds = {
    "South": [],
    "Midwest": [],
    "West": [],
    "Northeast": [],
}

for row in rows:
    new_ds[row['region']].append(row['obesity_pct'])


result = []
for region in new_ds.keys():
    result.append(f"{region} ~{compute_mean(new_ds, region)}%")

print(', '.join(result))
```
