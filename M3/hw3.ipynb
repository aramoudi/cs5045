````python
import pandas as pd

# Load the dataset. Each row becomes a Python dictionary; all 16 rows form a list.
df = pd.read_csv('data.csv')
students = df.to_dict('records')

print(f'Loaded {len(students)} student records.')
print('Columns:', list(students[0].keys()))
````

---
## Section 1: Categorize with if / elif / else

**Operation: Transform.** Produce a new label for a single numeric input based on a threshold rule.

**Rule:**
- `attendance_pct >= 85` → `'high'`
- `attendance_pct >= 70` → `'average'`
- `attendance_pct < 70`  → `'low'`

Fill in the six blanks below (two per sub-question).

````python
# 1a.
attendance = 88
if attendance >= 85:
    category = 'high'
elif attendance >= 70:
    category = 'average'
else:
    category = 'low'
print('1a.', category)        # expected: high

# 1b.
attendance2 = 72
if attendance2 >= 85:
    category2 = 'high'
elif attendance2 >= 70:
    category2 = 'average'
else:
    category2 = 'low'
print('1b.', category2)       # expected: average

# 1c.
attendance3 = 63
if attendance3 >= 85:
    category3 = 'high'
elif attendance3 >= 70:
    category3 = 'average'
else:
    category3 = 'low'
print('1c.', category3)       # expected: low
````

---
## Section 2: Compound conditions: and, or, not

Combining conditions with `and`, `or`, and `not` adds precision to the rule; it does not change the operation type.

````python
# 2a. 'Good standing': attendance_pct >= 85 AND assignment_avg >= 85.
att_test  = 90
asgn_test = 87
if att_test >= 85 and asgn_test >= 85:
    standing = 'good standing'
else:
    standing = 'not good standing'
print('2a.', standing)         # expected: good standing

# 2b. 'At risk': attendance_pct < 75 OR assignment_avg < 70.
att_test2  = 70
asgn_test2 = 80
if att_test2 < 75 or asgn_test2 < 70:
    risk = 'at risk'
else:
    risk = 'on track'
print('2b.', risk)             # expected: at risk

# 2c. 'Needs support': did NOT pass the midterm.
passed_flag = 'no'
if not passed_flag == 'yes':
    support = 'needs support'
else:
    support = 'on track'
print('2c.', support)          # expected: needs support
````

---
## Section 3: Use a for loop to process every row

A loop lets you apply the same operation to all 16 students. Combining a loop with a conditional gives you either a Filter (row count changes) or a Transform (same row count, new values).

````python
# 3a. Print each student's name and final_score.
print('--- 3a. Student Final Scores ---')
for student in students:
    print(student['name'], student['final_score'])    # fill in: access the student's name
````

````python
# 3b. Count how many students have attendance_pct >= 85.
count_high = 0
for student in students:
    if student['attendance_pct'] >= 85:          # condition: attendance_pct >= 85
        count_high += 1
print('3b. Students with attendance >= 85:', count_high)    # expected: 8
````

````python
# 3c. Collect the names of all Sociology majors.
sociology_names = []
for student in students:
    if student['major'] == 'Sociology':          # condition: major == 'Sociology'
        sociology_names.append(student['name'])  # append the student's name
print('3c. Sociology students:', sociology_names)
# expected: ['Dana Park', 'Jenna Ramirez', 'Omar Singh']
````

---
## Section 4: Implement the Filter operation

**Pattern:** empty list → loop through all rows → `append` only the rows that satisfy the condition. The row count decreases.

````python
# 4a. Keep only students who passed the midterm (passed_midterm == 'yes').
passed = []
for student in students:
    if student['passed_midterm'] == 'yes':
        passed.append(student)
print('4a. Students who passed the midterm:', len(passed))   # expected: 10
````

````python
# 4b. Keep only Biology majors.
biology = []
for student in students:
    if student['major'] == 'Biology':
        biology.append(student)                               # append student
print('4b. Biology students:', len(biology))                 # expected: 3
````

````python
# 4c. Keep students who passed the midterm AND have attendance_pct >= 75.
good_standing = []
for student in students:
    if student['passed_midterm'] == 'yes' and student['attendance_pct'] >= 75:
        good_standing.append(student)
print('4c. Passed midterm AND attendance >= 75:', len(good_standing))  # expected: 10
````

````python
# 4d. Keep students with final_score below 70 (struggling).
struggling = []
for student in students:
    if student['final_score'] < 70:
        struggling.append(student)
print('4d. Struggling students (final < 70):', len(struggling))        # expected: 6
````

---
## Section 5: Implement the Transform operation

**Pattern:** loop through all rows → compute a new value → store it back in the row's dictionary. The row count stays the same; each row gains a new key.

````python
# 5a. Add 'attendance_level' to every student.
#     Rule: 'high' (>= 85), 'average' (>= 70), 'low' (< 70)
for student in students:
    if student['attendance_pct'] >= 85:
        student['attendance_level'] = 'high'
    elif student['attendance_pct'] >= 70:
        student['attendance_level'] = 'average'
    else:
        student['attendance_level'] = 'low'

print('--- 5a. Attendance Levels (first 5) ---')
for student in students[:5]:
    print(student['name'], '->', student['attendance_level'])
# expected:
#   Aisha Nkemdirim  -> high
#   Carlos Vega      -> average
#   Dana Park        -> high
#   Evan Morse       -> low
#   Fatima Al-Rashid -> high
````

````python
# 5b. Compute a composite score for every student.
#     composite = 0.3 * midterm_score + 0.5 * final_score + 0.2 * assignment_avg
for student in students:
    student['composite_score'] = 0.3 * student['midterm_score'] + 0.5 * student['final_score'] + 0.2 * student['assignment_avg']

print('--- 5b. Composite Scores (first 5) ---')
for student in students[:5]:
    print(student['name'], '->', round(student['composite_score'], 1))
# expected (approximately):
#   Aisha Nkemdirim  -> 84.4
#   Carlos Vega      -> 67.8
#   Dana Park        -> 93.6
#   Evan Morse       -> 52.4
#   Fatima Al-Rashid -> 82.0
````
